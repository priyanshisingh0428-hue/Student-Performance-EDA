```python

# STUDENT PERFORMANCE ANALYSIS USING EDA

# Step 1: Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Step 2: Load Dataset
df = pd.read_csv("student-mat.csv", sep=";")

print("STUDENT PERFORMANCE DATASET")

# Step 3: Display Dataset
print("\nFirst Five Rows:")
print(df.head())

print("\nLast Five Rows:")
print(df.tail())

print("\nDataset Shape:")
print(df.shape)

print("\nDataset Information:")
df.info()

print("\nStatistical Description:")
print(df.describe())

# Step 4: Check Missing Values
print("\nMissing Values:")
print(df.isnull().sum())

# Step 5: Check Duplicate Values
print("\nDuplicate Records:")
print(df.duplicated().sum())

# Step 6: Average Grades Using NumPy
g1 = df["G1"].to_numpy()
g2 = df["G2"].to_numpy()
g3 = df["G3"].to_numpy()

print("\nAVERAGE GRADES")
print("Average G1:", np.mean(g1))
print("Average G2:", np.mean(g2))
print("Average G3:", np.mean(g3))

print("\nHighest Final Grade:", np.max(g3))
print("Lowest Final Grade:", np.min(g3))
print("Median Final Grade:", np.median(g3))

# Step 7: Grade Categories
def grade_category(grade):
    if grade >= 15:
        return "Excellent"
    elif grade >= 10:
        return "Good"
    elif grade >= 5:
        return "Average"
    else:
        return "Needs Improvement"

df["Grade_Category"] = df["G3"].apply(grade_category)

print("\nStudents in Each Grade Category:")
print(df["Grade_Category"].value_counts())

# Step 8: Average Grades by Gender
print("\nAverage Final Grade by Gender:")
print(df.groupby("sex")["G3"].mean())

# Step 9: Average Grades by Study Time
print("\nAverage Final Grade by Study Time:")
print(df.groupby("studytime")["G3"].mean())

# Step 10: Students with Highest Absences
print("\nTop 10 Students with Highest Absences:")
print(df.sort_values(by="absences", ascending=False)
      [["sex", "age", "absences", "G3"]].head(10))

# Step 11: Filtering Students
print("\nStudents with Final Grade Above 15:")
print(df[df["G3"] > 15][["sex", "studytime", "G3"]])

# Step 12: Grade Histogram
plt.figure(figsize=(8, 5))
plt.hist(df["G3"], bins=10, edgecolor="black")
plt.title("Distribution of Final Grades")
plt.xlabel("Final Grade (G3)")
plt.ylabel("Number of Students")
plt.savefig("grade_histogram.png")
plt.show()

# Step 13: Study Time Count Chart
plt.figure(figsize=(8, 5))
df["studytime"].value_counts().sort_index().plot(
    kind="bar", edgecolor="black"
)
plt.title("Number of Students by Study Time")
plt.xlabel("Study Time Category")
plt.ylabel("Number of Students")
plt.xticks(rotation=0)
plt.savefig("study_time_chart.png")
plt.show()

# Step 14: Three Observations
print("\nPROJECT OBSERVATIONS")

print("1. G1, G2 and G3 show students' academic performance across three periods.")

print("2. Average final grades can be compared across gender and study-time categories.")

print("3. Absences vary among students and can be examined alongside final grades.")

print("\nPROJECT COMPLETED SUCCESSFULLY!")
```
